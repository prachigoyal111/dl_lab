```python
import numpy as np

X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([[0],[1],[1],[0]])

np.random.seed(0)
W1 = np.random.randn(2,2)

W2 = np.random.randn(2,1)

def sigmoid(x):
    return 1/(1+np.exp(-x))

def sigmoid_deriv(x):
    return x*(1-x)

for i in range(10000):

    z1 = X.dot(W1)
    a1 = sigmoid(z1)

    z2 = a1.dot(W2)
    a2 = sigmoid(z2)

    # loss gradient
    d2 = (a2 - y) * sigmoid_deriv(a2)

    dW2 = a1.T.dot(d2)

    d1 = d2.dot(W2.T) * sigmoid_deriv(a1)

    dW1 = X.T.dot(d1)

    W1 -= 0.1 * dW1
    W2 -= 0.1 * dW2

print(np.round(a2))
```
